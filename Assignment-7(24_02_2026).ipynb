#### Dataset Detective 
Load a dataset, display top rows, find highest value column, count missing values, share 5 insights. Objective • Load a dataset • Display top rows • Find the column with highest value • Count missing values • Provide 5 insights

```python
student_data = { "Student_ID": [101,102,103,104,105,106,107,108,109,110], 
                 "Name": ["Ravi","Priya","Rahul","Anu","Kiran","Sneha","Arjun","Meera","Dev", "Avi"] ,
                "Age": [20,21,20,22,21,20,23,21,22,20], 
                "Gender": ["Male","Female","Male","Female","Male","Female","Male","Female", "Male", "Female"], 
                "Math_Score": [85,92,70,60,45,88,76,95,68,82],
                "Science_Score": [78,88,65,58,50,84,72,91,70,79], 
                "English_Score": [82,90,72,65,55,86,74,94,None,85], 
                "Attendance_Percentage": [92,95,85,80,70,93,88,97,82,90], 
                "Study_Hours_Per_Day": [3,4,2,2,1,3,2,5,2,3] }
```

```python
import pandas as pd 
# Load dataset 
df = pd.DataFrame(student_data) 
# Load dataset if u have csv file 
#df = pd.read_csv("data.csv") 
# Display top rows 
print("Top 5 Rows:") 
print(df.head())
```

```python
# Find highest value column 
print("\nMaximum values in each column:") 
print(df.max(numeric_only=True))
```

```python
# Column with overall highest value 
highest_value = df.max(numeric_only=True) 
highest_column = df.max(numeric_only=True).idxmax() 
print("\nColumn with highest value:", highest_column) 
print("Highest value:", highest_value)
```

```python
# Count missing values 
print("\nMissing values in each column:") 
print(df.isnull().sum())
```

```python
## 5 insights 
top_math = df.loc[df["Math_Score"].idxmax()] 
print("Top Math Student:") 
print(top_math["Name"], top_math["Math_Score"])
```

```python
max_values = df.max(numeric_only=True) 
print("Maximum value in each column:") 
print(max_values) 
print("Column with highest value:", max_values.idxmax())
```

```python
missing_values = df["English_Score"].isnull().sum() 
print("Missing values in English_Score:", missing_values)
```

```python
students_range = df[(df["Math_Score"] >= 70) & (df["Math_Score"] <= 90)] 
print("Students scoring between 70 and 90 in Math:") 
print(students_range[["Name","Math_Score"]])
```
